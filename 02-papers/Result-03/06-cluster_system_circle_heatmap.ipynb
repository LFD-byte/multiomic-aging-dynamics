```python
from pathlib import Path
import importlib
import sys
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

_candidates = [Path.cwd().resolve(), Path.cwd().resolve() / '02-papers' / 'Result-03']
HERE = next(path for path in _candidates if (path / 'plot_cluster_system_circle_heatmap.py').exists())
if str(HERE) not in sys.path:
    sys.path.insert(0, str(HERE))
import plot_cluster_system_circle_heatmap as circle_heatmap
plot_circle_heatmap = importlib.reload(circle_heatmap).plot_circle_heatmap
```

```python
ASSOCIATION_FILE = HERE / 'outputs' / 'cluster_system_association' / 'cluster_system_associations.csv'
OUTPUT_PREFIX = HERE / 'figures' / 'cluster_system_circle_heatmap'
FIGURE_WIDTH_MM = 210
FIGSIZE = (FIGURE_WIDTH_MM / 25.4, FIGURE_WIDTH_MM / 25.4)
# The rings end at 90° (straight up); the opening is to the right of the end boundary.
# Cluster IDs sit just right of the end boundary, aligned with the rings; the colour bar is to their right; no title or subtitle.
# Outer labels read outside-in on the left half and inside-out on the right half.

# At most the Top 10 phenotypes per system; all are shown when a system has fewer than 10.
# Ranked by smallest FDR, then by largest absolute partial_r. Set to 0 to show all.
TOP_FEATURES_PER_SYSTEM = 10
FDR_COLUMN = 'fdr_global'
FDR_THRESHOLD = 0.05
VLIM = None  # e.g. 0.6 for a fixed -0.6 to 0.6 colour scale
ASSOCIATION_FILE
```

```python
associations = pd.read_csv(ASSOCIATION_FILE)
print('Rows:', len(associations))
print('Systems:', associations['system'].nunique())
print('Phenotypes:', associations[['system', 'feature']].drop_duplicates().shape[0])
print('Clusters:', sorted(associations['cluster_id'].unique()))
associations.head()
```

```python
fig, feature_order = plot_circle_heatmap(
    associations,
    effect_col='partial_r',
    fdr_col=FDR_COLUMN,
    fdr_threshold=FDR_THRESHOLD,
    top_features_per_system=TOP_FEATURES_PER_SYSTEM,
    vlim=VLIM,
    figsize=FIGSIZE,
    output_prefix=OUTPUT_PREFIX,
)
display(fig)
plt.close(fig)
print(f'Saved: {OUTPUT_PREFIX}.png, {OUTPUT_PREFIX}.pdf and {OUTPUT_PREFIX}.svg')
```

# Covariate Adjustment

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import warnings
warnings.filterwarnings('ignore')

# ══════════════════════════════════════════════════════════════
# 1. Configuration
# ══════════════════════════════════════════════════════════════
META_COLS = ['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_date', 
             'collection_timestamp', 'timezone', 'year_of_birth', 'month_of_birth', 
             'collection_datetime', 'batch', 'data_source', 'hmo', 'bmi', 'sex', 'age']

CONFOUNDERS  = ['bmi', 'sex']   # confounders to regress out
AGE_MIN, AGE_MAX = 40, 70
MIN_SAMPLES=20

MICROBIOME = {'Gut Microbiome', 'Oral Microbiome'}

# Omics data
OMICS = {
    'Transcriptomics': df_rna_with_metadata,
    # 'Metabolomics':    df_metabo_with_metadata,
    # 'Lipidomics':      df_lipid_with_metadata,
    # 'Gut Microbiome':  df_gut_with_metadata,
    # 'Oral Microbiome': df_oral_with_metadata,
}

# ══════════════════════════════════════════════════════════════
# 2. CLR transform (microbiome)
# ══════════════════════════════════════════════════════════════
def clr_transform(X: np.ndarray) -> np.ndarray:
    X = X.astype(float) + 1.0
    X = X / X.sum(axis=1, keepdims=True)
    X = np.log(X)
    return X - X.mean(axis=1, keepdims=True)

# ══════════════════════════════════════════════════════════════
# 3. Regress confounders out of each feature and keep the residuals
# ══════════════════════════════════════════════════════════════
def regress_out_confounders(df: pd.DataFrame,
                            feat_cols: list,
                            confounders: list) -> pd.DataFrame:
    """
    Fit an independent linear regression for each feature column:
        feature ~ bmi + sex
    and return the residuals as adjusted values.
    """
    avail_conf = [c for c in confounders if c in df.columns]
    if not avail_conf:
        print("  Warning: Confounder columns not found, skipping regression correction")
        return df[feat_cols].copy()

    cov_mat = df[avail_conf].values.astype(float)
    resid_df = pd.DataFrame(index=df.index, columns=feat_cols, dtype=float)

    for feat in feat_cols:
        y    = df[feat].values.astype(float)
        mask = ~(np.isnan(y) | np.any(np.isnan(cov_mat), axis=1))
        if mask.sum() < MIN_SAMPLES:
            resid_df[feat] = np.nan
            continue
        reg = LinearRegression().fit(cov_mat[mask], y[mask])
        resid         = np.full(len(y), np.nan)
        resid[mask]   = y[mask] - reg.predict(cov_mat[mask])
        resid_df[feat] = resid

    return resid_df

# ══════════════════════════════════════════════════════════════
# 4. Main workflow: process all omics layers
# ══════════════════════════════════════════════════════════════
print("Step 1: Regression correction + age profile computation...")

omics_adjusted = {}

for oname, df_raw in OMICS.items():
    print(f"  [{oname}]", end=' ')
    df = df_raw.copy()
    df['participant_id'] = df['participant_id'].astype(str)

    # Filter by age range
    df = df[(df['age'] >= AGE_MIN) & (df['age'] <= AGE_MAX)].copy()
    df = df.dropna(subset=['age'])

    feat_cols = [c for c in df.columns if c not in META_COLS]
    metadata_cols = [c for c in df.columns if c in META_COLS]

    # CLR transform for microbiome layers
    X = df[feat_cols].values.astype(float)
    if oname in MICROBIOME:
        X = clr_transform(X)
        df[feat_cols] = X

    # Drop low-variance features
    col_std   = np.nanstd(X, axis=0)
    valid_idx = col_std > 1e-10
    feat_cols = [f for f, v in zip(feat_cols, valid_idx) if v]

    # Keep metadata
    meta_df = df[metadata_cols].copy()

    # Build the data used for regression
    df_model = df[['participant_id', 'age'] + CONFOUNDERS + feat_cols].copy()
    df_model = df_model.dropna(subset=CONFOUNDERS)
    # Ensure sex is numeric (0/1)
    df_model['sex'] = pd.to_numeric(df_model['sex'], errors='coerce')

    # Align indices
    meta_df = meta_df.loc[df_model.index]

    # Confounder adjustment
    resid_df  = regress_out_confounders(df_model, feat_cols, CONFOUNDERS)

    # Concatenate metadata and residuals
    resid_with_meta = pd.concat([meta_df, resid_df], axis=1)

    print(resid_with_meta.shape)

    omics_adjusted[oname] = resid_with_meta

omics_adjusted['Transcriptomics'].to_parquet(f'{OUTPUT_PREFIX}/transcriptomics_adjusted.parquet', index=False)
# omics_adjusted['Metabolomics'].to_parquet(f'{OUTPUT_PREFIX}/metabolomics_adjusted.parquet', index=False)
# omics_adjusted['Lipidomics'].to_parquet(f'{OUTPUT_PREFIX}/lipidomics_adjusted.parquet', index=False)
# omics_adjusted['Gut Microbiome'].to_parquet(f'{OUTPUT_PREFIX}/gut_microbiome_adjusted.parquet', index=False)
# omics_adjusted['Oral Microbiome'].to_parquet(f'{OUTPUT_PREFIX}/oral_microbiome_adjusted.parquet', index=False)

omics_adjusted['Transcriptomics'].to_csv(f'{OUTPUT_PREFIX}/transcriptomics_adjusted.csv', index=False)
# omics_adjusted['Metabolomics'].to_csv(f'{OUTPUT_PREFIX}/metabolomics_adjusted.csv', index=False)
# omics_adjusted['Lipidomics'].to_csv(f'{OUTPUT_PREFIX}/lipidomics_adjusted.csv', index=False)
# omics_adjusted['Gut Microbiome'].to_csv(f'{OUTPUT_PREFIX}/gut_microbiome_adjusted.csv', index=False)
# omics_adjusted['Oral Microbiome'].to_csv(f'{OUTPUT_PREFIX}/oral_microbiome_adjusted.csv', index=False)
```

# Data View

```python
import pandas as pd
import sys
sys.path.append("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/phenoAI/")

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/vascular_health/vascular_health.parquet'
df_pvh = read_parquet(pfile)
print(df_pvh.shape)
df_pvh.head()
```

```python
missing_analyzer = MissingValueAnalyzer(df_pvh)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```

# Missing Value Handling

```python
metadata_col = ['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_date', 'collection_timestamp', 'timezone']
df_pvh_metadata = df_pvh[metadata_col]
df_pvh_features = df_pvh.drop(columns=metadata_col)
```

```python
means = df_pvh_features.mean()
df_pvh_imputed = df_pvh_features.fillna(means)
```

# Save

```python
df_pvh_metadata.reset_index(drop=True, inplace=True)
df_pvh_imputed.reset_index(drop=True, inplace=True)
df_pvh = pd.concat([df_pvh_metadata, df_pvh_imputed], axis=1)
pfile = 'pheno_data/202512/010-peripheral_vascular_health_imputed.parquet'
df_pvh.to_parquet(pfile, index=False)
```

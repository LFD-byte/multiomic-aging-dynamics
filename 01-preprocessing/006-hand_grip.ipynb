# Data View

```python
import pandas as pd
import sys
sys.path.append("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/phenoAI/")

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/hand_grip/hand_grip.parquet'
df_hand_grip = read_parquet(pfile)
print(df_hand_grip.shape)
df_hand_grip.head()
```

```python
missing_analyzer = MissingValueAnalyzer(df_hand_grip)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```

# Data filtering

```python
metadata_col = ['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_date', 'timezone', 'dominant_hand']
feature_col = ['hand_grip_strength_left', 'hand_grip_strength_right']
metadata_col = [col for col in metadata_col if col in df_hand_grip.columns]
df_hand_grip_metadata = df_hand_grip[metadata_col]
df_hand_grip_features = df_hand_grip[feature_col]

mask_nota = df_hand_grip_features.notna().all(axis=1)
df_hand_grip_metadata = df_hand_grip_metadata.loc[mask_nota]
df_hand_grip_features = df_hand_grip_features.loc[mask_nota]
```

# Save

```python
df_hand_grip_metadata.reset_index(drop=True, inplace=True)
df_hand_grip_features.reset_index(drop=True, inplace=True)
df_hand_grip = pd.concat([df_hand_grip_metadata, df_hand_grip_features], axis=1).dropna(axis=1)
pfile = 'pheno_data/202512/006-hand_grip_imputed.parquet'
df_hand_grip.to_parquet(pfile, index=False)
```

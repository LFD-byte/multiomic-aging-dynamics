# Data View

```python
import pandas as pd
import sys
sys.path.append("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/phenoAI/")

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/blood_pressure/blood_pressure.parquet'
df_bpress = read_parquet(pfile)
print(df_bpress.shape)
df_bpress.head()
```

```python
missing_analyzer = MissingValueAnalyzer(df_bpress)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```

# Data filtering

```python
metadata_col = ['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_date', 'collection_timestamp', 'timezone', 'dizziness']
no_need_col = ['sitting_second_blood_pressure_diastolic', 'sitting_second_blood_pressure_pulse_rate', 'sitting_second_blood_pressure_systolic', 'blood_pressure_hand']
df_bpress_metadata = df_bpress[metadata_col]
df_bpress_features = df_bpress.drop(columns=metadata_col + no_need_col)
```

```python
mask_nota = df_bpress_features.notna().all(axis=1)
df_bpress_metadata = df_bpress_metadata.loc[mask_nota]
df_bpress_features = df_bpress_features.loc[mask_nota]
```

# Save

```python
df_bpress_metadata.reset_index(drop=True, inplace=True)
df_bpress_features.reset_index(drop=True, inplace=True)
df_bpress = pd.concat([df_bpress_metadata, df_bpress_features], axis=1)
pfile = 'pheno_data/202512/007-blood_pressure_imputed.parquet'
df_bpress.to_parquet(pfile, index=False)
```

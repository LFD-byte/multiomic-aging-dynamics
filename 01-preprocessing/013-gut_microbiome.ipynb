## Data View

```python
import sys
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_DIR = Path("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/Project/2026-nc-final").resolve()
DATASETS_DIR = (Path("/home/ec2-user/studies") / "hpp_datasets").resolve()

sys.path.append(str(PROJECT_DIR))

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = f'{DATASETS_DIR}/gut_microbiome/gut_microbiome.parquet'
df_gut = read_parquet(pfile)
print(df_gut.shape)
df_gut.head()
```

```python
pfile = f'{DATASETS_DIR}/gut_microbiome/metaphlan4/aggregated/metaphlan_abundance_species.parquet'
df_gut_metap_species = read_parquet(pfile)
print(df_gut_metap_species.shape)
df_gut_metap_species.head()
```

```python
df_gut = pd.merge(df_gut_metap_species, 
                      df_gut[['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_timestamp', 'collection_date']], 
                      on=['participant_id', 'cohort', 'research_stage', 'array_index'])
df_gut = df_gut.dropna(subset=['participant_id', 'cohort', 'research_stage', 'array_index'])
print(df_gut.shape)
df_gut.head()
```

```python
# Check the number and proportion of missing values
missing_analyzer = MissingValueAnalyzer(df_gut)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```

## Missing Value Handling

```python
gut_species_metadata_col = ['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_date', 'collection_timestamp', 'timezone']
gut_species_metadata_col = [col for col in gut_species_metadata_col if col in df_gut.columns]
df_gut_metadata = df_gut[gut_species_metadata_col]
df_gut_features = df_gut.drop(columns=gut_species_metadata_col)
```

## Imputation

```python
impute_value = 0.0001
min_samples = 500

# detected = (df_gut_features > 0) # 0 和 nan 为 false
detected = ~df_gut_features.isna() # nan 为 false，0 视为有意义值
detected_counts = detected.sum(axis=0)  # 每列（物种）被检测到的样本数
keep_species = detected_counts[detected_counts >= min_samples].index
df_gut_features_filtered = df_gut_features.loc[:, keep_species].copy()

df_gut_features_imputed = df_gut_features_filtered.fillna(impute_value)
print(df_gut_features_imputed.shape)
df_gut_features_imputed.head()
```

```python
df_gut_metadata.reset_index(drop=True, inplace=True)
df_gut = pd.concat([df_gut_metadata, df_gut_features_imputed], axis=1)
df_gut.to_parquet('processed_data/013-gut_microbiome_imputed.parquet')
```

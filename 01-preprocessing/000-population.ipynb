```python
import pandas as pd
import sys
sys.path.append("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/phenoAI/")

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/population/population.parquet'
df_popu = read_parquet(pfile)
print(df_popu.shape)
df_popu.head()
```

```python
missing_analyzer = MissingValueAnalyzer(df_popu)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```

```python
# 0: female, 1: male
rate_female = df_popu[df_popu['sex'] == '0'].shape[0] / df_popu.shape[0]
rate_male = 1 - rate_female
print(f"male: {rate_male}, female: {rate_female}")
```

# Data View

```python
import pandas as pd
import sys
sys.path.append("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/phenoAI/")

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/carotid_ultrasound/carotid_ultrasound.parquet'
df_carotid = read_parquet(pfile)
print(df_carotid.shape)
df_carotid.head()
```

```python
missing_analyzer = MissingValueAnalyzer(df_carotid)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```

# Missing Value Handling

```python
carotid_metadata_col = ['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_date', 'collection_timestamp', 'timezone']
carotid_metadata_col = [col for col in carotid_metadata_col if col in df_carotid.columns]
df_carotid_metadata = df_carotid[carotid_metadata_col]
df_carotid_features = df_carotid.drop(columns=carotid_metadata_col)
```

```python
means = df_carotid_features.mean()
df_carotid_imputed = df_carotid_features.fillna(means)
```

# Save

```python
df_carotid_metadata.reset_index(drop=True, inplace=True)
df_carotid_imputed.reset_index(drop=True, inplace=True)
df_carotid = pd.concat([df_carotid_metadata, df_carotid_imputed], axis=1).dropna(axis=1)
pfile = 'pheno_data/202512/012-carotid_ultrasound_imputed.parquet'
df_carotid.to_parquet(pfile, index=False)
```

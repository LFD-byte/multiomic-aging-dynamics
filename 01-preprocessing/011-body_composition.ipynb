# Data View

```python
import pandas as pd
import sys
sys.path.append("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/phenoAI/")

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/body_composition/body_composition.parquet'
df_body_comp = read_parquet(pfile)
print(df_body_comp.shape)
df_body_comp.head()
```

```python
missing_analyzer = MissingValueAnalyzer(df_body_comp)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```

# Missing Value Handling

```python
body_comp_metadata_col = ['participant_id', 'cohort', 'research_stage', 'array_index', 'collection_date', 'timezone', 'hmo']
cols_rm = ['collection_timestamp']
body_comp_metadata_col = [col for col in body_comp_metadata_col if col in df_body_comp.columns]
df_body_comp_metadata = df_body_comp[body_comp_metadata_col]
df_body_comp_features = df_body_comp.drop(columns=body_comp_metadata_col + cols_rm)
```

```python
means = df_body_comp_features.mean()
df_body_comp_imputed = df_body_comp_features.fillna(means)
```

# Save

```python
df_body_comp_metadata.reset_index(drop=True, inplace=True)
df_body_comp_imputed.reset_index(drop=True, inplace=True)
df_body_comp = pd.concat([df_body_comp_metadata, df_body_comp_imputed], axis=1).dropna(axis=1)
pfile = 'pheno_data/202512/011-body_composition_imputed.parquet'
df_body_comp.to_parquet(pfile, index=False)
```

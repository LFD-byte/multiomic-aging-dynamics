# Data View

```python
import pandas as pd
import sys
sys.path.append("/home/ec2-user/studies/HUIFA-RNA-SEQ-new/phenoAI/")

from phenokit.preprocessing.qc import MissingValueAnalyzer
from phenokit.preprocessing.io import read_parquet
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/fundus/fundus.parquet'
df_fundus = read_parquet(pfile)
print(df_fundus.shape)
df_fundus.head()
```

```python
pfile = '/home/ec2-user/studies/hpp_datasets/fundus/microvasculature.parquet'
df_fundus_mvas = read_parquet(pfile)
print(df_fundus_mvas.shape)
df_fundus_mvas.head()
```

```python
df_fundus_mvas = pd.merge(df_fundus_mvas, df_fundus[['participant_id', 'research_stage', 'array_index', 'collection_date']], on=['participant_id', 'research_stage', 'array_index'], how='left')
```

```python
df_fundus_metadata = pd.read_csv('/home/ec2-user/studies/hpp_datasets/fundus/metadata/fundus_data_dictionary.csv')
need_feature = df_fundus_metadata[df_fundus_metadata['field_type'].isin(['Continuous', 'Categorical (Single)', 'Datetime', 'Date',])]['tabular_field_name']
df_fundus_metadata.head()
```

```python
df_fundus_mvas = df_fundus_mvas.loc[:, df_fundus_mvas.columns.isin(need_feature.tolist() + ['participant_id', 'cohort', 'research_stage', 'array_index'])]
```

```python
missing_analyzer = MissingValueAnalyzer(df_fundus_mvas)
missing_analyzer.overall_missing_rate()
missing_analyzer.report_features()
missing_analyzer.report_samples()
missing_analyzer.report_heatmap()
```
